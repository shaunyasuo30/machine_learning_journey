# Series in Pandas

```python
import pandas as pd
import numpy as np
```

```python
data1 = pd.Series([1,2,3,4,5], name = 'numbers')
data1
```

```python
data = pd.Series(['C++', 'Golang', 'Java', 'Python', 'Swift'],
                 index = list('CGJPS'),
                 name = 'Programming Language')
print(data)
```

```python
for v in data:
  print(v)
```

```python
print(data[2:3])
print(data['G':'J'])
print(data1.loc[2:3])
print(data[['P', 'C']])
print(data1.between(3,6))
result = data1[data1.between(3,6)]
print(result)
```

```python
result = data.drop('C', inplace=False)
print(data)
print(result)

data.drop('C', inplace=True)
data
```

```python
data['K'] = 'Kotlin'
data
```

```python
data.sort_index(inplace=True)
data
```

```python
data = pd.Series([5,4,6,3,1,6], name = 'Numbers')
data[2.5] = 19
data.sort_index(inplace=True)
print(data)
data = data.reset_index(drop=True)
print(data)
```

```python
print(data.min())
print(data.max())
print(data.argmax())
print(data.idxmax())
print(data.sum())
print(data.mean())
print(data.std())
print(data.var())
```

```python
data = pd.Series(['C++', 'Golang', 'Java', 'Python', 'Swift'],
                 index = list('CGJPS'),
                 name = 'Programming Language')
print(data.str.count("Java"))
print(data.str.count('a'))
```

```python
test = data.str.upper()
print(test, '\n')
test1 = data.replace('C++', 'C#')
print(test1)
```

```python
center1 = pd.Series([12, 23, 31, 11, 9],
                    index=list(['C++', 'Golang',
                                'Java', 'Python',
                                'Swift']),
                    name='num_registered')

center2 = pd.Series([18, 44, 49, 27],
                    index=list(['Golang', 'Java',
                                'Python', 'Swift']),
                    name='num_registered')

total = center1 + center2
print(total)
```

```python
data = pd.Series([4, 1, 0, 2, 2, 1],
                 index=list(['C++', 'Golang',
                             'Java', 'Golang',
                             'Java', 'C++']),
                 name='num_dropped')
print(data, '\n')
print(data['C++'].sum())
print(data['Golang'].sum())
print(data['Java'].sum())
```

**1D Interpolation**

```python
data = pd.Series([1, 6, 3, 8, np.nan, 7, np.nan, 2],
                 name='num_dropped')
data.interpolate(method='linear', inplace=True)
data
```

#Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.jointplot(x = 'sepal_length', y = 'sepal_width', data = data, hue = 'species')
plt.show()
```

```python
sns.jointplot(x = 'petal_length', y = 'petal_width', data = data, hue = 'species')
plt.show()
```

```python
sns.stripplot(x='species', y='petal_length', data=data, hue='species')
plt.show()
```

```python
sns.stripplot(x='species', y='sepal_length', data=data, hue='species')
plt.show()
```

```python
sns.boxplot(x='species', y='sepal_length', data=data, hue='species')
plt.show()
```
