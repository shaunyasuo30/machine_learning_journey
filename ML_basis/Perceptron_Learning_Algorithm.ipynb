Là một thuật toán Classification

$J_1(\mathbf{w}) = \sum_{\mathbf{x}_i \in \mathcal{M}} (-y_i\text{sgn}(\mathbf{w}^T\mathbf{x_i}))
$ </h>

Hàm đếm số lượng điểm missclassifed. Khi hàm này = 0 thì sẽ ko còn điểm nào missclassifed nữa

Tuy nhiên, đây là hàm rời rạc, không tính được đạo hàm theo **w** nên rất khó tối ưu. Ta cần tìm 1 hàm khác để việc tối ưu khả khi hơn


$J(w)=∑xi∈M(−yiw^Txi)$ <br>
Hàm J()khác một chút với hàm J1() ở việc bỏ đi hàm sgn. Nhận xét rằng khi một điểm misclassified xi nằm càng xa boundary thì giá trị −yiwTxi sẽ càng lớn, nghĩa là sự sai lệch càng lớn. Giá trị nhỏ nhất của hàm mất mát này cũng bằng 0 nếu không có điểm nào bị misclassifed. Hàm mất mát này cũng được cho là tốt hơn hàm J1() vì nó trừng phạt rất nặng những điểm lấn sâu sang lãnh thổ của class kia. Trong khi đó, J1() trừng phạt các điểm misclassified như nhau (đều = 1), bất kể chúng xa hay gần với đường biên giới.


Dùng SGD (Stochastic Gradient Descent) để tối ưu hàm mất mát này.

Với một điểm dữ liệu xi bị misclassified, hàm mất mát trở thành:

$J(\mathbf{w}; \mathbf{x}_i; y_i) = -y_i\mathbf{w}^T\mathbf{x}_i
$

Đạo hàm tương ứng:

$J(\mathbf{w}; \mathbf{x}_i; y_i) = -y_i\mathbf{w}^T\mathbf{x}_i
$

Vậy quy tắc cập nhật là:

$\mathbf{w} = \mathbf{w} + \eta y_i\mathbf{x}_i
$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import cdist
np.random.seed(23)

means = [[2, 2], [4, 2]]
cov = [[.3, .2], [.2, .3]]
N = 10
X0 = np.random.multivariate_normal(means[0], cov, N).T
X1 = np.random.multivariate_normal(means[1], cov, N).T

X = np.concatenate((X0, X1), axis = 1)
y = np.concatenate((np.ones((1, N)), -1*np.ones((1, N))), axis = 1)
# Xbar
X = np.concatenate((np.ones((1, 2*N)), X), axis = 0)
```

```python
def h(w,x):
  return np.sign(np.dot(w.T,x))
def has_converged(X,y,w):
  return np.array_equal(h(w,X),y)
def perceptron(X,y,w_init):
  w = [w_init]
  N = X.shape[1]
  d = X.shape[0]
  mis_points = []
  while True:
    mix_id = np.random.permutation(N)
    for i in range(N):
      xi = X[:,mix_id[i]].reshape(d,1)
      yi = y[0,mix_id[i]]
      if h(w[-1],xi)[0] != yi:
        mis_points.append(int(mix_id[i]))
        w_new = w[-1] + yi*xi
        w.append(w_new)
    if has_converged(X,y,w[-1]):
      break
  return (w,mis_points)

d = X.shape[0]
w_init = np.random.randn(d, 1)
(w,miss_points) = perceptron(X, y, w_init)
print(w[-1])
print(miss_points)
```
